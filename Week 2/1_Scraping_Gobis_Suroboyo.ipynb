# **Scrapping Google Play Review: Gobis Suroboyo**

### Prosedur Pengumpulan Data (Data Collection)**

Tahap pengumpulan data dilakukan dengan metode **web scraping** pada platform **Google Play Store** untuk mengambil ulasan pengguna aplikasi **Gobis Suroboyo**. Proses ekstraksi data diimplementasikan menggunakan pustaka **`google-play-scraper`** melalui fungsi **`reviews_all`**. Parameter pencarian dikonfigurasi secara spesifik pada paket aplikasi **`id.go.surabaya.gobis`** dengan pengaturan bahasa. Seluruh ulasan yang tersedia diambil dengan metode pengurutan berdasarkan ulasan terbaru (**`Sort.NEWEST`**) guna memastikan dataset yang diperoleh tetap relevan dengan kondisi terkini aplikasi. Langkah ini bertujuan untuk membangun basis data primer yang objektif untuk kemudian diproses pada tahapan analisis selanjutnya.

### **Libraries**

```python
!pip install google_play_scraper

from google_play_scraper import reviews_all, Sort
import pandas as pd
import numpy as np
```

### **Scraping Gobis Suroboyo App Review**

```python
# Mengambil ulasan dari play store (bahasa inggris)
gobis_reviews = reviews_all(
    'id.go.surabaya.gobis',
    sleep_milliseconds=0,  # defaults to 0
    lang='en',  # defaults to 'en'
    sort=Sort.NEWEST,  # defaults to Sort.MOST_RELEVANT
)
```

### **Simpan ulasan ke dalam Dataframe**

```python
# Simpan ulasan ke dalam DataFrame
df_gobis = pd.DataFrame(np.array(gobis_reviews), columns=['content'])
df_gobis = df_gobis.join(pd.DataFrame(df_gobis.pop('content').tolist()))

df_gobis
```

### **Simpan Dataframe ke file CSV**

```python
# Simpan DataFrame ke file CSV
output_path = "rawdata_gobis.csv"
df_gobis.to_csv(output_path, index=False, encoding="utf-8")
```
