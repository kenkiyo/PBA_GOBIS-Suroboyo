```python
corpus = [
    'Pemerintah Kota Surabaya terus berinovasi dalam mengintegrasikan layanan transportasi publik.',
    'Aplikasi Gobis Suroboyo menjadi tulang punggung digitalisasi transportasi massal.',
    'Inisiatif ini mendorong warga beralih ke transportasi umum dan mengurangi emisi karbon.'
]

corpus
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
cv.fit(corpus)
print(cv.vocabulary_)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Inisialisasi CountVectorizer
cv = CountVectorizer()

# Fit + transform corpus
independentFeatures = cv.fit_transform(corpus).toarray()
independentFeatures
```
