```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
# Mendefinisikan corpus berupa list dari 3 dokumen berdasarkan artikel di atas
corpus = [
    'Pemerintah Kota Surabaya terus berinovasi mengintegrasikan layanan transportasi publik untuk kenyamanan warga.',
    'Aplikasi Gobis Suroboyo membantu masyarakat melacak armada bus secara real-time dan mendukung transportasi massal.',
    'Inovasi Suroboyo Bus yang ramah lingkungan bertujuan untuk mengurangi emisi karbon di pusat kota Surabaya.'
]

corpus
```

```python
# Inisialisasi CountVectorizer
vectorizer = CountVectorizer()

# Fit dan transform corpus menjadi matriks fitur
X = vectorizer.fit_transform(corpus)

# Menampilkan Feature Matrix (Term Frequency)
df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
print("Feature Matrix (Term Frequency):")
df
```

```python
full_cos_sim_matrix = cosine_similarity(X)

# Menampilkan matriks dalam bentuk DataFrame agar mudah dibaca
cos_sim_df = pd.DataFrame(full_cos_sim_matrix, 
                          index=[f'Doc {i}' for i in range(len(corpus))], 
                          columns=[f'Doc {i}' for i in range(len(corpus))])

print("Full Cosine Similarity Matrix:")
cos_sim_df
```
